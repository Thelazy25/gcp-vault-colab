Notebook ini adalah **Tier 1** dari Kerangka Kerja Render Bertingkat GCP Vault.
- **Biaya**: Rp 0 (100% Gratis - Free Tier Google Colab)
- **Hardware**: NVIDIA Tesla T4 GPU (16 GB VRAM) dengan akselerasi hardware NVENC
- **Tujuan**: Prototyping, uji coba filter FFmpeg, AI video generation, dan render draf tanpa memakan saldo GCP.

```python
# 1. Periksa GPU Tesla T4 & Status Akselerasi Hardware
!nvidia-smi
!apt-get update -qq && apt-get install -y -qq ffmpeg fonts-dejavu-core
```

```python
# 2. Siapkan Folder Kerja Lokal Colab
import os
os.makedirs('/content/input', exist_ok=True)
os.makedirs('/content/output', exist_ok=True)
print('Folder siap: /content/input dan /content/output')
```

```python
# 3. Eksekusi Render Video Berakselerasi GPU NVENC (1080x1920 60fps 9:16)
# Menggunakan encoder hardware NVIDIA (h264_nvenc) untuk kecepatan render 3x - 5x lebih cepat dari CPU
!ffmpeg -y \
    -f lavfi -i testsrc=duration=10:size=1080x1920:rate=60 \
    -f lavfi -i sine=frequency=440:duration=10 \
    -vf "drawtext=fontfile=/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf:text='GCP VAULT TIER 1 - COLAB GPU T4':fontcolor=white:fontsize=44:x=(w-text_w)/2:y=(h-text_h)/2-100:box=1:boxcolor=black@0.6:boxborderw=10, drawtext=fontfile=/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf:text='GPU Accelerated Render':fontcolor=cyan:fontsize=36:x=(w-text_w)/2:y=(h-text_h)/2+20:box=1:boxcolor=black@0.6:boxborderw=10" \
    -c:v h264_nvenc -preset p5 -tune hq -b:v 6M -pix_fmt yuv420p \
    -c:a aac -b:a 192k \
    /content/output/render_tier1_colab.mp4

print('\n[SUKSES] Render GPU selesai!')
!ls -lh /content/output
```
